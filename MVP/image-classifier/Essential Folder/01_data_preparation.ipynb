```python
import os
import torch
import torchvision
import torchvision.transforms as transforms
```

```python
os.listdir('./Dataset/')
```

```python
training_dataset_path = './archive/training/training/'
```

```python
training_transforms = transforms.Compose([
    transforms.Resize((224, 224)), 
    transforms.ToTensor()
])

# Process images at 224,224 pixels
```

```python
train_dataset = torchvision.datasets.ImageFolder(root = training_dataset_path, transform = training_transforms )
# Scnas folder to look for labels
```

```python
train_loader = torch.utils.data.DataLoader(dataset = train_dataset, batch_size=32, shuffle = False)
# Delivers them to the neural network, dataset is the all the pics, take in batch of 32, and does not mix them 
```

```python
def get_mean_and_std(loader):
    mean = 0.
    std = 0.
    total_images_count = 0
    for images,_ in loader:
        images_count_in_a_batch = images.size(0) 
        images = images.view(images_count_in_a_batch, images.size(1), -1)# takes all 2d img in a batch and flattens them to make python calculate the math fast
        mean += images.mean(2).sum(0) # avg colour for rgb
        std += images.std(2).sum(0) # avg brightness
        total_images_count += images_count_in_a_batch
    
    mean /= total_images_count
    std /= total_images_count
    return mean, std

# Calculates avg brightness(mean) and colour (deviation).. It is to normalise data so that AI can learn faster
```

```python
get_mean_and_std(train_loader)
```
